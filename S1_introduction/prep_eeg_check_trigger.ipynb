```python
# Import necessary libraries
import mne
```

```python
# Read a .bdf file from the specified path, which contains EEG data
raw = mne.io.read_raw_bdf("C:/Users/mfbpe/Desktop/DATA/2023_Eva_Freedom/raw\\Pa1.bdf")

# Print the raw data object to view its summary, which includes details like
# the number of channels, the sampling frequency, and the duration of the recording
print(raw)
```

```python
# Access and display the list of channel names to verify that they match the expected ones
raw.info["ch_names"]
```

```python
# Find and detect events (triggers) in the data, with a minimum event duration of 1 sample
# This helps to identify event markers that indicate specific actions or stimuli during the recording
events= mne.find_events(raw, shortest_event=1)
print(events)
```

```python
# Set up an interactive matplotlib window to visually plot the detected events
%matplotlib qt
mne.viz.plot_events(events) 
```

```python
raw.plot(events=events, scalings=dict(eeg=200e-6))
```
